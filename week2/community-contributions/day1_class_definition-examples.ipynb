```python
# Test streaming with gradio
import gradio as gr
```

```python
def stream_model(systemPrompt, prompt, model):
    if model=="GPT (gpt-4o-mini)":
        llm = GPT_Wrapper(systemPrompt, prompt)
    elif model=="Ollama (llama3.2)":
        llm = Ollama_Wrapper(systemPrompt, prompt)
    elif model=="Claude (claude-3-5-haiku-20241022)":
        llm = Claude_Wrapper(systemPrompt, prompt)
    elif model=="Gemini (gemini-1.5-flash)":
        llm = Gemini_Wrapper(systemPrompt, prompt)
    else:
        raise ValueError("Unknown model")
    result = llm.getResultStream()
    yield from result
    
```

```python
view = gr.Interface(
    fn=stream_model,
    inputs=[
        gr.Textbox(label="System prompt:", value=system_msg), 
        gr.Textbox(label="Your message:", value=user_msg), 
        gr.Dropdown([
            "GPT (gpt-4o-mini)", 
            "Ollama (llama3.2)", 
            "Claude (claude-3-5-haiku-20241022)", 
            "Gemini (gemini-1.5-flash)"
        ], 
        label="Select model", 
        value="Ollama (llama3.2)")],
    outputs=[gr.Markdown(label="Response:")],
    flagging_mode="never"
)
view.launch()
```
